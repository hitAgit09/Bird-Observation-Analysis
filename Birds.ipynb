```python
import pandas as pd

# File names (must match uploaded names exactly)
forest_file = "Bird_Monitoring_Data_FOREST.XLSX"
grass_file = "Bird_Monitoring_Data_GRASSLAND.XLSX"

# Read Excel files
forest_xls = pd.ExcelFile(forest_file)
grass_xls = pd.ExcelFile(grass_file)

df_list = []

# 🔹 FOREST data
for sheet in forest_xls.sheet_names:
    temp = pd.read_excel(forest_xls, sheet_name=sheet)
    temp["Location_Type"] = "Forest"  # Add habitat
    temp["Admin_Unit_Code"] = sheet   # Track sheet
    df_list.append(temp)

# 🔹 GRASSLAND data
for sheet in grass_xls.sheet_names:
    temp = pd.read_excel(grass_xls, sheet_name=sheet)
    temp["Location_Type"] = "Grassland"
    temp["Admin_Unit_Code"] = sheet
    df_list.append(temp)

# 🔹 Combine all sheets
df = pd.concat(df_list, ignore_index=True)

# Preview
df.head()
```

```python
df.shape
df.info()
df.isnull().sum()
```

# **DATA CLEANING**

```python
df["Temperature"] = df["Temperature"].fillna(df["Temperature"].mean())
df["Humidity"] = df["Humidity"].fillna(df["Humidity"].median())
```

```python
df.dropna(subset=["Scientific_Name"], inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
cols_to_drop = ["AcceptedTSN", "NPSTaxonCode", "AOU_Code"]
df.drop(columns=cols_to_drop, inplace=True, errors='ignore')
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors='coerce')

df["Start_Time"] = pd.to_datetime(df["Start_Time"], errors='coerce')
df["End_Time"] = pd.to_datetime(df["End_Time"], errors='coerce')
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Hour"] = df["Start_Time"].dt.hour

# Season column
def get_season(m):
    if m in [12,1,2]: return "Winter"
    elif m in [3,4,5]: return "Summer"
    elif m in [6,7,8]: return "Monsoon"
    else: return "Autumn"

df["Season"] = df["Month"].apply(get_season)

# Observation duration
df["Observation_Duration"] = (df["End_Time"] - df["Start_Time"]).dt.seconds / 60
```

```python
df.to_csv("/content/cleaned_bird_data.csv", index=False)
```

# **SQL DATABASE**

```python
engine = create_engine("sqlite:///birds.db")

df.to_sql("bird_data", engine, if_exists="replace", index=False)
```

## **SQL Queries**

## Top Species

```python
pd.read_sql("""
SELECT Common_Name, COUNT(*) as count
FROM bird_data
GROUP BY Common_Name
ORDER BY count DESC
LIMIT 10;
""", engine)
```

## Habitat Diversity

```python
pd.read_sql("""
SELECT Location_Type, COUNT(DISTINCT Scientific_Name) as species_count
FROM bird_data
GROUP BY Location_Type;
""", engine)
```

## Seasonal Trends

```python
pd.read_sql("""
SELECT Season, COUNT(*) as observations
FROM bird_data
GROUP BY Season;
""", engine)
```

## At-Risk Species

```python
pd.read_sql("""
SELECT Common_Name, COUNT(*)
FROM bird_data
WHERE PIF_Watchlist_Status = 'TRUE'
GROUP BY Common_Name;
""", engine)
```
